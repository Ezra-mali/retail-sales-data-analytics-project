# Data Cleaning

## Project: Retail Sales Data Analytics

This notebook focuses on preparing the dataset for analysis by addressing data quality issues and ensuring consistency.

The key steps include:
- Handling missing values
- Fixing data types
- Removing duplicates
- Preparing a clean dataset for further analysis

```python
import pandas as pd

df = pd.read_csv("../data/raw/Sample - Superstore.csv", encoding="latin1")
df.head()
```

```python
df.isnull().sum()
```

## Missing Values Handling

The dataset was checked for missing values across all columns.

No missing values were found, indicating that the dataset is complete and does not require imputation or removal of records at this stage.

This ensures that subsequent analysis will not be affected by incomplete data.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df[['Order Date', 'Ship Date']].dtypes
```

## Data Type Conversion

The `Order Date` and `Ship Date` columns were originally stored as object (string) data types.

These columns were converted to datetime format to enable time based analysis, such as trend analysis, time series evaluation, and feature extraction (e.g., year, month).

This ensures that date related operations can be performed accurately in later stages of the project.

```python
df.duplicated().sum()
```

## Duplicate Records Handling

The dataset was checked for duplicate records using the `duplicated()` function.

No duplicate rows were found in the dataset, indicating that each transaction is unique.

Therefore, no records were removed at this stage.

```python
(df['Sales'] < 0).sum()
```

## Sales Consistency Check

The dataset was checked for negative values in the Sales column.

No negative values were found, indicating that all recorded sales amounts are valid and consistent with expected business rules.

This confirms that the Sales data does not require correction at this stage.

```python
(df['Profit'] < 0).sum()
```

## Profit Consistency Check

The dataset was checked for negative values in the Profit column.

A total of 1871 transactions were found to have negative profit values.

Unlike Sales, negative Profit values are valid and indicate loss-making transactions. These may result from factors such as high discounts, low pricing, or high costs.

These records were retained, as they provide important insights into business performance and will be further analyzed in later stages of the project.

```python
df.to_csv("../data/processed/cleaned_superstore.csv", index=False)
```

## Summary

This stage prepared the dataset for analysis by ensuring data quality, consistency, and feature availability.
