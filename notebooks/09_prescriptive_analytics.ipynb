# Prescriptive Analytics

Descriptive analytics identified historical business performance, diagnostic analytics explained the factors contributing to those results, and predictive analytics explored the use of historical data to estimate future sales.

The purpose of prescriptive analytics is to recommend practical business actions based on the insights obtained from previous analyses. These recommendations are intended to help decision makers improve profitability, optimise operations and support long-term business growth.

## Business Objective

The objective of this notebook is to develop data-driven business recommendations that address the key issues identified throughout the descriptive, diagnostic and predictive analyses.

The recommendations focus on improving profitability, reducing losses, supporting better decision making and enabling sustainable business performance.

## Summary of Previous Findings

The previous stages of the project identified several important business insights:

- Technology generated the highest sales and profit.
- Furniture generated the lowest overall profit despite achieving relatively high sales.
- Furniture received the highest average discount among all product categories.
- Tables and Bookcases were the primary contributors to losses within the Furniture category.
- The Central region experienced losses in the Furniture category, with Furnishings, Tables and Bookcases performing particularly poorly.
- The predictive models demonstrated limited forecasting accuracy, indicating that additional business variables are required to improve future sales predictions.

## Recommendation 1: Review Furniture Pricing and Discount Strategy

The analysis showed that the Furniture category generated the lowest overall profit while receiving the highest average discount. In addition, Tables and Bookcases recorded negative profits, suggesting that the current pricing and discount strategy may be reducing profitability.

The business should review its discount policies for Furniture products, particularly Tables and Bookcases, to determine whether discounts can be reduced without significantly affecting sales. Product pricing and supplier costs should also be evaluated to improve profit margins.

Implementing a more balanced pricing strategy could increase profitability while maintaining competitive product offerings.

## Recommendation 2: Prioritise High-Performing Product Categories

Technology consistently generated the highest sales and profit during the analysis period, making it the strongest-performing product category.

The business should continue investing in Technology products through targeted marketing campaigns, adequate inventory levels and product expansion. Maintaining the availability of high performing products can help sustain revenue growth and customer satisfaction.

Focusing on products with strong financial performance enables the business to maximise returns on its investments and strengthen its competitive position.

## Recommendation 3: Improve the Performance of Underperforming Regions

The analysis showed that the Central region recorded losses within the Furniture category, particularly for Tables, Bookcases and Furnishings.

The business should investigate the factors contributing to these losses by reviewing regional pricing strategies, customer demand, inventory management and operational costs. Regional managers should also assess whether promotional campaigns or product offerings need to be adjusted to improve financial performance.

Addressing the causes of poor regional performance can help reduce losses and improve overall profitability.

## Recommendation 4: Strengthen Data Collection and Sales Forecasting

The predictive analysis demonstrated that the available data was insufficient to produce highly accurate sales forecasts using a simple linear regression model.

The business should continue collecting high quality historical sales data and incorporate additional variables such as promotional activities, customer purchasing behaviour, product pricing and seasonal events into future predictive models.

Improving data quality and predictive modelling capabilities will enable the business to make more accurate forecasts, optimise inventory planning and support better strategic decision-making.

## Expected Business Impact

Implementing the recommendations presented in this notebook could improve both the financial and operational performance of the business. Reviewing pricing and discount strategies may increase profitability within underperforming product categories, while continued investment in high performing categories can support revenue growth.

Investigating regional performance and strengthening sales forecasting capabilities will enable the business to make more informed decisions regarding inventory management, marketing activities and resource allocation. Collectively, these actions can improve operational efficiency, enhance profitability and support long-term business success.

# Prescriptive Analytics Summary

This notebook translated the findings from the descriptive, diagnostic and predictive analyses into practical business recommendations. Each recommendation was supported by evidence obtained from the analysis of historical retail sales data.

The recommendations focused on improving profitability, strengthening high-performing product categories, addressing regional performance issues and enhancing future sales forecasting. Together, these actions provide a structured approach for improving business performance through data-driven decision making.

This completes the analytics lifecycle for the retail sales project by progressing from understanding historical performance, identifying underlying causes, estimating future outcomes and recommending actions that support better business decisions.
