# Feature Engineering

## Project: Retail Sales Data Analytics

This notebook focuses on creating new features from the cleaned dataset to enhance analysis and support deeper business insights.

The key steps include:
- Creating time-based features
- Deriving business metrics
- Preparing data for advanced analytics and visualization

```python
import pandas as pd

df = pd.read_csv("../data/processed/cleaned_superstore.csv")
df.head()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year

df[['Order Date', 'Year']].head()
```

```python
df['Month'] = df['Order Date'].dt.month

df[['Order Date', 'Month']].head()
```

```python
df['Month Name'] = df['Order Date'].dt.month_name()

df[['Order Date', 'Month', 'Month Name']].head()
```

```python
df['Profit Margin'] = df['Profit'] / df['Sales']

df[['Sales', 'Profit', 'Profit Margin']].head()
```

```python
df['Customer Segment Type'] = pd.qcut(df['Sales'], q=3, labels=['Low', 'Medium', 'High'])

df[['Sales', 'Customer Segment Type']].head()
```

## Feature Engineering Summary

The following features were created to enhance the dataset and support deeper analysis:

- **Year**: Extracted from Order Date to enable yearly trend analysis.
- **Month**: Numeric month extracted for time-based grouping and sorting.
- **Month Name**: Provides readable month labels for reporting and visualization.
- **Profit Margin**: Calculated as Profit divided by Sales to measure profitability efficiency.
- **Customer Segment Type**: Created using sales quantiles to classify transactions into Low, Medium, and High value segments.

These features improve the dataset’s analytical value and support more meaningful insights in subsequent analysis and visualization stages.

```python
df.to_csv("../data/processed/feature_engineered_superstore.csv", index=False)
```
